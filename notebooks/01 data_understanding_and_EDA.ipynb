# Feature analysis using a reusable function

```python
def analyze_numerical_feature(df, feature, target):
    """
    Analyze a numerical feature using:
    1. Histogram
    2. Scatter plot with target variable
    """

    # Create a figure
    plt.figure(figsize=(12, 5))

    # ---------------- Histogram ----------------
    plt.subplot(1, 2, 1)
    plt.hist(df[feature], bins=30)
    plt.title(f"Distribution of {feature}")
    plt.xlabel(feature)
    plt.ylabel("Frequency")

    # ---------------- Scatter Plot ----------------
    plt.subplot(1, 2, 2)
    plt.scatter(df[feature], df[target], alpha=0.5)
    plt.title(f"{feature} vs {target}")
    plt.xlabel(feature)
    plt.ylabel(target)

    plt.tight_layout()
    plt.show()
```

```python
top_features = [
    "Overall Qual",
    "Gr Liv Area",
    "Garage Cars",
    "Garage Area",
    "Total Bsmt SF",
    "1st Flr SF",
    "Year Built",
    "Full Bath",
    "Year Remod/Add",
    "Garage Yr Blt"
]
```

```python
for feature in top_features:
    analyze_numerical_feature(
        df,
        feature,
        "SalePrice"
    )
```

# Important Categorical Feature based on Observations

```python
top_categorical_features = [
    "Neighborhood",
    "House Style",
    "Bldg Type",
    "Kitchen Qual",
    "Exter Qual",
    "Foundation",
    "Garage Type",
    "Sale Condition",
    "MS Zoning",
    "Roof Style"
]
```

```python
### Reusable function for categorical features
import matplotlib.pyplot as plt

def analyze_categorical_feature(df, feature, target):

    print("=" * 50)
    print(f"Feature: {feature}")
    print("=" * 50)

    print(f"Missing Values : {df[feature].isnull().sum()}")
    print(f"Unique Values  : {df[feature].nunique()}")
    print(f"Most Common    : {df[feature].mode()[0]}")

    plt.figure(figsize=(14,5))

    # Frequency Chart
    plt.subplot(1,2,1)
    df[feature].value_counts().plot(kind="bar")
    plt.title(f"{feature} Distribution")
    plt.xticks(rotation=90)

    # Box Plot
    plt.subplot(1,2,2)
    df.boxplot(column=target, by=feature, rot=90)

    plt.title(f"{target} by {feature}")
    plt.suptitle("")
    plt.tight_layout()

    plt.show()
```

```python
for feature in top_categorical_features:
    analyze_categorical_feature(
        df,
        feature,
        "SalePrice"
    )
```

## 📍 Location

Houses located in certain neighborhoods consistently have higher selling prices, highlighting that location is one of the strongest factors influencing house value.

## ⭐ Construction Quality

Better kitchen and exterior quality are associated with higher selling prices, indicating that construction quality significantly impacts property value.


## 🏠 Building Characteristics

Building type and house style influence selling price, although their impact is less pronounced than overall quality or location.


## 🚗 Garage

Houses with attached garages generally have higher selling prices compared with detached or absent garages.

## 📄 Sale Characteristics

Sale condition affects the selling price, with normal sales generally exhibiting more stable price distributions than special sale conditions.
